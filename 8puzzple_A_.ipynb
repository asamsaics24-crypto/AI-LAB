```python
import heapq

def manhattan(state, goal):
    distance = 0

    for i in range(9):
        if state[i] != '0':
            pos = goal.index(state[i])
            distance += abs(i // 3 - pos // 3) + abs(i % 3 - pos % 3)

    return distance

def neighbors(state):
    result = []
    zero = state.index('0')
    row = zero // 3
    col = zero % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        nr = row + dr
        nc = col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_zero = nr * 3 + nc
            temp = list(state)
            temp[zero], temp[new_zero] = temp[new_zero], temp[zero]
            result.append(''.join(temp))

    return result

def astar(start, goal):
    pq = []
    heapq.heappush(pq, (manhattan(start, goal), 0, start, [start]))
    visited = set()

    while pq:
        f, g, state, path = heapq.heappop(pq)

        if state == goal:
            return path

        if state in visited:
            continue

        visited.add(state)

        for next_state in neighbors(state):
            if next_state not in visited:
                new_g = g + 1
                new_f = new_g + manhattan(next_state, goal)
                heapq.heappush(pq, (new_f, new_g, next_state, path + [next_state]))

    return None

def display(state):
    print(state[0], state[1], state[2])
    print(state[3], state[4], state[5])
    print(state[6], state[7], state[8])

print("8-Puzzle using A* Algorithm")
print("Enter 0 for the blank space")

start = input("Enter initial state: ")
goal = input("Enter goal state: ")

path = astar(start, goal)

if path:
    print("\nSolution Path\n")

    for i, state in enumerate(path):
        print("Step", i)
        display(state)

        if i < len(path) - 1:
            print("   ↓")

    print("\nTotal Moves:", len(path) - 1)

else:
    print("\nNo solution exists")
```
