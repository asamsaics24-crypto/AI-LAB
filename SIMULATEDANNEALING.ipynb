```python
import random
import math
def calculate_cost(board):
    conflicts = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts
def generate_neighbor(board):
    new_board = board.copy()

    col = random.randint(0, 7)
    new_row = random.randint(0, 7)

    while new_row == new_board[col]:
        new_row = random.randint(0, 7)

    new_board[col] = new_row

    return new_board

def simulated_annealing():
    n = 8
    board = [random.randint(0, 7) for _ in range(n)]

    temperature = 100.0
    cooling_rate = 0.995
    min_temperature = 0.001

    current_cost = calculate_cost(board)

    while temperature > min_temperature:

        if current_cost == 0:
            return board
        neighbor = generate_neighbor(board)
        neighbor_cost = calculate_cost(neighbor)

        delta = neighbor_cost - current_cost
        if delta < 0:
            board = neighbor
            current_cost = neighbor_cost
        elif random.random() < math.exp(-delta / temperature):
            board = neighbor
            current_cost = neighbor_cost
        temperature *= cooling_rate

    if current_cost == 0:
        return board

    return None

def display_board(board):
    for row in range(8):
        for col in range(8):
            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()

solution = simulated_annealing()

if solution:
    print("Solution Found!")
    print("Queen positions:", solution)
    print("\nChessboard:")
    display_board(solution)
    print("Attacking pairs:", calculate_cost(solution))
else:
    print("No solution found. Try running again.")
print("\n1WN24CS052")
print("\nASAM SAI NISHITH REDDY")
```
